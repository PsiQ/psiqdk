# How to Write a Simple Program

This how-to guide walks you through the high-level steps of writing a simple Workbench program. 

You can read more about each step of the process in the corresponding tutorial.

## Step 1. Initialize a QPU

A Workbench program starts by creating a [QPU object](../tutorials/QPU-Object.ipynb) - the representation of the quantum processing unit on which this program will run. This object specifies how the program will be compiled and executed.

```python
qpu = QPU(num_qubits=4)
```

## Step 2. Allocate qubits

The next step is allocating the qubits for the program to use. Workbench represents qubit registers as either "raw" qubits ([Qubits data type](../tutorials/Qubits-Data-Type.ipynb)) or [arithmetic data types](../tutorials/Quantum-Arithmetic-Data-Types.ipynb).

```python
reg1 = Qubits(2, "reg1", qpu=qpu)
reg2 = Qubits(2, "reg2", qpu=qpu)
```

## Step 3. Perform the computation

The next step is implementing the logic of the computation the program needs to do. In Workbench, there are three main ways of doing this:

1. For simple programs, you can apply [primitive gates](../tutorials/Basic-Gates.ipynb) directly using the methods of the Qubits data type.

   ```python
   reg1[0].had()
   reg1[1].x(cond=reg1[0])
   ```

2. For more complicated logic, you can use [built-in library routines in Workbench](../tutorials/Built-in-Qubricks.ipynb) and in [Workbench Algorithms](https://construct.psiquantum.com/docs/psiqdk-algorithms/index.html).

   ```python
   state_prep = ArbitraryStatePrep([1, 0, 0, 1])
   state_prep.compute(reg2)
   ```

3. You can also [write your own routines](../tutorials/Qubricks.ipynb) to implement custom logic.

## Step 4. Get the results

The final step is getting the results using [measurements](../tutorials/Measurements.ipynb). 

```python
res1 = reg1.read()
res2 = reg2.read()
```

## Example

The following example shows a Workbench program that prepares two Bell states in two different ways (using the primitive gates and using state preparation library) and then measures them.

````python
from psiqdk.workbench import QPU, Qubits
from psiqdk.algorithms import ArbitraryStatePrep

# Initialize a QPU
qpu = QPU(num_qubits=4)

# Allocate qubits
reg1 = Qubits(2, "reg1", qpu=qpu)
reg2 = Qubits(2, "reg2", qpu=qpu)

# Perform the computation using primitive gates
reg1[0].had()
reg1[1].x(cond=reg1[0])

# Perform the computation using library routines
state_prep = ArbitraryStatePrep([1, 0, 0, 1])
state_prep.compute(reg2)

# Get the results
res1 = reg1.read()
res2 = reg2.read()
print("Measurement results:", res1, res2)
````
